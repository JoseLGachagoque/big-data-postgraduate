## 2. Diagnóstico de cinco campos · 10 minutos

Construir una tabla llamada `perfil` para `Secuencial`, `Fecha de Análisis`, `Departamento`, `Municipio` y `pH agua:suelo`. Para cada campo se registran los nulos de pandas, los textos vacíos después de retirar espacios y los marcadores `ND`. Las tres situaciones se cuentan por separado.

📘 **Apoyo:** `isna`, `.str.strip`, `.str.upper`, `.eq` y `sum`.

✅ **Comprobación:** `perfil` debe tener una fila por campo. Con `keep_default_na=False`, un texto vacío o `ND` no se convierte automáticamente en un nulo.

````python
# Desarrollo de la actividad 2
# Crear y mostrar perfil con las cinco columnas solicitadas.
````

````python
# Campos a analizar
campos = ["Secuencial", "Fecha de Análisis", "Departamento", "Municipio", "pH agua:suelo"]

# Construcción del perfil
perfil = pd.DataFrame([
    {
        "Campo": col,
        "Nulos": suelo[col].isna().sum(),
        "Vacíos": (suelo[col].str.strip() == "").sum(),
        "ND": suelo[col].str.strip().str.upper().eq("ND").sum()
    }
    for col in campos
])

# Mostrar perfil
display(perfil)
````

**Interpretación del resultado:**

_Pendiente de completar con evidencia de las tablas o controles anteriores._

perfil tiene una fila por campo.

Los tres tipos de ausencia se cuentan por separado.

Los textos vacíos y “ND” no se convierten automáticamente en nulos, lo cual queda reflejado en las columnas correspondientes.


Nulos: son los valores que pandas reconoce como NaN.

Vacíos: son cadenas que tras retirar espacios, quedaron en blanco ("").

ND: son registros explícitos con el marcador "ND", que pandas no convierte en nulo porque se usó keep_default_na=False.

Cada fila del perfil resume la calidad de los datos de ese campo: cuántos registros faltan, cuántos están vacíos y cuántos fueron marcados como “ND”.

Esto permite identificar problemas de completitud y decidir estrategias de limpieza (ej. imputar, descartar o transformar).

## 3. Conversión de pH y fecha · 15 minutos

Crear `ph` desde `pH agua:suelo` y `fecha_dt` desde `Fecha de Análisis`. Utilizar conversión numérica con `errors="coerce"` y fecha con formato `%d/%m/%Y`. Conservar los originales. Distinguir entradas vacías de textos no vacíos que no pudieron convertirse.

📘 **Apoyo:** `pd.to_numeric`, `.str.replace`, `pd.to_datetime`, `isna` y máscaras booleanas.

✅ **Comprobación:** Se muestran los conteos de conversiones fallidas y hasta cinco ejemplos con su valor original. No se reemplazan medidas ausentes por cero.

````python
# Desarrollo de la actividad 3
# Crear suelo["ph"] y suelo["fecha_dt"].
# Mostrar los fallos de conversión y sus valores originales.

# Crear columna ph con conversión numérica
suelo["ph"] = pd.to_numeric(
    suelo["pH agua:suelo"].str.replace(",", ".", regex=False),  # reemplazar coma decimal
    errors="coerce"
)

# Crear columna fecha_dt con conversión a datetime
suelo["fecha_dt"] = pd.to_datetime(
    suelo["Fecha de Análisis"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Detectar fallos de conversión en ph
mask_ph_fail = suelo["ph"].isna() & suelo["pH agua:suelo"].str.strip().ne("")
print("Fallos en ph:", mask_ph_fail.sum())
print(suelo.loc[mask_ph_fail, ["pH agua:suelo"]].head())

# Detectar fallos de conversión en fecha_dt
mask_fecha_fail = suelo["fecha_dt"].isna() & suelo["Fecha de Análisis"].str.strip().ne("")
print("Fallos en fecha_dt:", mask_fecha_fail.sum())
print(suelo.loc[mask_fecha_fail, ["Fecha de Análisis"]].head())
````

**Interpretación del resultado:**

_Pendiente de completar con evidencia de las tablas o controles anteriores._

suelo[ph] contendrá valores numéricos o NaN cuando la conversión falle.

suelo[fecha_dt] contendrá objetos datetime64 o NaT cuando la conversión falle.

Los conteos (mask_ph_fail.sum() y mask_fecha_fail.sum()) mostrarán cuántos registros no pudieron convertirse aunque no estaban vacíos.


Entradas vacías → se distinguen porque la cadena original está vacía (""), y no se consideran fallos de conversión.

Fallos de conversión → son textos no vacíos que no pudieron transformarse en número o fecha con el formato indicado.

Esto permite diferenciar entre ausencia de datos y errores de formato o contenido.

Conservar los originales (pH agua:suelo y Fecha de Análisis) asegura que siempre se pueda revisar el valor original frente al convertido.

No se reemplazan medidas ausentes por cero: los NaN y NaT se mantienen para reflejar la falta de información o el error de conversión.

## 4. Registros para revisión · 15 minutos

Crear la columna booleana `en_revision`. Se marca una fila si su pH es ausente, está fuera de `[0, 14]` o su fecha no se pudo interpretar. Contar por separado cada motivo y el total de filas marcadas. Conservar todas las observaciones.

📘 **Apoyo:** `between`, `isna`, `|`, `~`, `.loc` y `value_counts`.

✅ **Comprobación:** Las filas marcadas y no marcadas deben sumar el total original. Una misma fila puede tener varios motivos; por eso la suma de los motivos no necesariamente coincide con el total marcado.

````python
# Desarrollo de la actividad 4
# Crear en_revision y los conteos por motivo.
# Comprobar la conservación del total original con assert.

# Crear columna booleana en_revision
suelo["en_revision"] = (
    suelo["ph"].isna() |                      # pH ausente
    ~suelo["ph"].between(0, 14) |             # pH fuera de rango
    suelo["fecha_dt"].isna()                  # fecha no interpretable
)

# Conteos por motivo
motivo_ph_ausente = suelo["ph"].isna().sum()
motivo_ph_fuera   = (~suelo["ph"].between(0, 14)).sum()
motivo_fecha_fail = suelo["fecha_dt"].isna().sum()

# Conteo total de filas marcadas
total_marcadas = suelo["en_revision"].sum()

print("Motivo: pH ausente ->", motivo_ph_ausente)
print("Motivo: pH fuera de [0,14] ->", motivo_ph_fuera)
print("Motivo: fecha no interpretable ->", motivo_fecha_fail)
print("Total filas marcadas en_revision ->", total_marcadas)

# Comprobación de conservación del total original
assert suelo.shape[0] == total_filas_original, "Se perdió alguna fila"
print("Comprobación OK: filas originales conservadas")
````

**Interpretación del resultado:**

_Pendiente de completar con evidencia de las tablas o controles anteriores._

en_revision será True en las filas que cumplan al menos uno de los criterios.

Los conteos por motivo mostrarán cuántas filas tienen pH ausente, cuántas están fuera de rango y cuántas tienen fecha no interpretable.

El total de filas marcadas (total_marcadas) será menor o igual a la suma de los motivos, porque una misma fila puede tener varios problemas.

El assert confirma que el número de filas sigue siendo 92.738, garantizando que no se eliminaron registros.


La columna en_revision funciona como una bandera de control de calidad: indica qué registros requieren revisión antes de análisis posteriores.

Los conteos separados permiten identificar la magnitud de cada problema (ausencia de pH, valores fuera de rango, fechas inválidas).

El total marcado refleja cuántas observaciones están afectadas, pero no coincide necesariamente con la suma de motivos porque una fila puede tener más de un error.

Conservar todas las observaciones asegura que no se pierda información: incluso los registros problemáticos se mantienen en la tabla, pero quedan señalados para tratamiento posterior.

## 5. Evidencias y conclusiones · 10 minutos

Construir `controles` con las columnas `control` y `cantidad`: total de entrada, pH sin número, pH fuera del intervalo, fecha no interpretable y total de filas en revisión. Exportar a `OUT / "controles_agrosavia.csv"` sin índice. Redactar dos hallazgos apoyados en los conteos.

📘 **Apoyo:** `pd.DataFrame`, `to_csv(index=False)` y las tablas de las actividades anteriores.

✅ **Comprobación:** El archivo se genera dentro de la ruta `OUT`. Cada hallazgo menciona un conteo y explica su significado o limitación.

````python
# Desarrollo de la actividad 5
# Construir controles, mostrarlo y exportarlo a OUT.
# Redactar los dos hallazgos en la celda Markdown siguiente.

# Construir tabla de controles
controles = pd.DataFrame([
    {"control": "Total de entrada", "cantidad": total_filas_original},
    {"control": "pH sin número", "cantidad": suelo["ph"].isna().sum()},
    {"control": "pH fuera del intervalo", "cantidad": (~suelo["ph"].between(0, 14)).sum()},
    {"control": "Fecha no interpretable", "cantidad": suelo["fecha_dt"].isna().sum()},
    {"control": "Total filas en revisión", "cantidad": suelo["en_revision"].sum()}
])

# Mostrar controles
display(controles)

# Exportar a CSV sin índice
controles.to_csv(OUT / "controles_agrosavia.csv", index=False)

# Comprobación de conservación del total original
assert suelo.shape[0] == total_filas_original, "Se perdió alguna fila"
print("Comprobación OK: filas originales conservadas")
````

**Interpretación del resultado:**

_Pendiente de completar con evidencia de las tablas o controles anteriores._

Hallazgos

Hallazgo 1: El conteo de pH sin número muestra que existen registros donde la medida no pudo convertirse a valor numérico. Esto indica problemas de calidad en la captura de datos, que deben revisarse antes de cualquier análisis químico.

Hallazgo 2: El hallazgo de pH fuera del intervalo [0 - 14] indica que existen registros con valores numéricos que no son físicamente posibles para una medida de pH.

El número de fechas no interpretables revelaría que hay inconsistencias en el formato de la columna “Fecha de Análisis”. Aunque no se eliminan las filas, estas limitaciones afectan la posibilidad de ordenar cronológicamente los registros y requieren limpieza o estandarización.
